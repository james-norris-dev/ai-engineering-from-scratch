# CSV Exploration\nLoad a dataset, view it, and plot a column

```python
import pandas as pd

df = pd.DataFrame({
    "day": range(1,8),
    "temp": [61, 63, 59, 65, 70, 68, 66]
})
df.to_csv("weather.csv", index=False)
df = pd.read_csv("weather.csv")
df
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

plt.plot(df["day"], df["temp"])
plt.title("Temp over the week")
plt.show()
```
